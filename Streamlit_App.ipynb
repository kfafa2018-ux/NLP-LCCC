```python
%%writefile app.py
import streamlit as st

# Title of the app
st.title('My First App')

# Setting up options & how to select them
options = ["1", "2", "3"]
choose = st.selectbox("Pick an option", options)

# Printing text to screen
st.write("Well done!")
st.write('Your first app!')
st.write("This app uses these models for classification: Naïve Bayes, SVM, Random Forest, Decision Tree")

# Creating Buttons
if st.button("Classify"):
    st.write("You have clicked the button")

# Sidebar section
st.sidebar.write("This is the sidebar. You can put things like buttons & checkboxes here!")
st.sidebar.button('New Button')
st.sidebar.selectbox("Pick an option:", options, key='sidebar_opt')
st.sidebar.checkbox('New Checkbox')
```
